## Figure 1 - Reaction model

**Author:** Quentin Duez

This notebook makes a numerical model of four consecutive equilibrium reactions in continuous flow conditions

**Outputs:** Figure 1A, Figure S1, Figure S2

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from scipy.integrate import odeint

sns.set_theme(style="ticks", context="paper")
```

```python
# Reaction network definition


def d4_rev_dt(x, t, *args):
    A, B, C, D, E = x
    k1, km1, k2, km2, k3, km3, k4, km4, kf = args
    dAdt = (
        -k1 * A + km1 * B + kf * A0_ss * (1 + np.sin(2 * np.pi * t / PERIOD)) - kf * A
    )
    dBdt = -km1 * B + k1 * A - k2 * B + km2 * C - kf * B
    dCdt = k2 * B - km2 * C - k3 * C + km3 * D - kf * C
    dDdt = k3 * C - km3 * D - k4 * D + km4 * E - kf * D
    dEdt = k4 * D - km4 * E - kf * E
    return dAdt, dBdt, dCdt, dDdt, dEdt


# Definition of parameters
A0_ss = 1  # mol/L
TAU = 30  # minutes - Residence time
PERIOD = 60  # min
kf = 1 / TAU  # = flowrate (µL/min)/Reactor volume (µL) = 1/residence time (min)
```

```python
# Four steps reaction : A <--> B <--> C <--> D <--> E
# With fixed k1, k2, k3, k4, km1, km2, km3, km4

t = np.linspace(0, 1000, 1001)
x0 = 1, 0, 0, 0, 0  # initial conditions
A_input = A0_ss * (1 + np.sin(2 * np.pi * t / PERIOD))

k1 = k2 = k3 = k4 = 0.1  # min-1
km1 = km2 = km3 = km4 = 0.001  # min-1

sol = odeint(d4_rev_dt, x0, t, args=(k1, km1, k2, km2, k3, km3, k4, km4, kf))

A_overlay = sol[:, 0] / sol[:, 0].mean()
B_overlay = sol[:, 1] / sol[:, 1].mean()
C_overlay = sol[:, 2] / sol[:, 2].mean()
D_overlay = sol[:, 3] / sol[:, 3].mean()
E_overlay = sol[:, 4] / sol[:, 4].mean()

fig, ax = plt.subplots(2, 1, figsize=(6, 2))
sns.lineplot(x=t, y=A_input, c="tab:blue", label="Input", linewidth=1.5, ax=ax[0])
sns.lineplot(x=t, y=A_overlay, c="tab:blue", label="A", linewidth=1.5, ax=ax[1])
sns.lineplot(x=t, y=B_overlay, c="tab:orange", label="B", linewidth=1.5, ax=ax[1])
sns.lineplot(x=t, y=C_overlay, c="tab:green", label="C", linewidth=1.5, ax=ax[1])
sns.lineplot(x=t, y=D_overlay, c="tab:red", label="D", linewidth=1.5, ax=ax[1])
sns.lineplot(x=t, y=E_overlay, c="tab:grey", label="E", linewidth=1.5, ax=ax[1])

# ax[0].set_ylim(0, 2)
ax[0].set_xlim(0, 300)
ax[1].set_ylim(0, 2)
ax[1].set_xlim(0, 300)

ax[0].set_ylabel("Input conc. (M)", fontsize=11)
ax[0].set_xlabel("Time (min)", fontsize=11)
ax[1].set_ylabel("Conc. / Avg.(Conc)", fontsize=11)
ax[1].set_xlabel("Time (min)", fontsize=11)

ax[0].legend(bbox_to_anchor=(1.02, 1.1), loc="upper left", fontsize=9)
ax[1].legend(bbox_to_anchor=(1.02, 1.1), loc="upper left", fontsize=9)

ax[0].set_facecolor("#F8F8FB")
ax[1].set_facecolor("#F8F8FB")
plt.subplots_adjust(
    left=None, bottom=None, right=None, top=1.2, wspace=None, hspace=0.55
)

# plt.savefig('Scheme_4_rev_reactions.png', format='png', bbox_inches='tight', dpi=600)
```

### SI : Other reaction connectivity

```python
def d4_rev_branched_dt(x, t, *args):
    A, B, C, D, E = x
    k1, km1, k2, km2, k3, km3, k4, km4, kf = args
    dAdt = (
        -k1 * A
        + km1 * B
        - k2 * A
        + km2 * C
        + kf * A0_ss * (1 + np.sin(2 * np.pi * t / PERIOD))
        - kf * A
    )
    dBdt = -km1 * B + k1 * A - kf * B
    dCdt = k2 * A - km2 * C - k3 * C + km3 * D - k4 * C + km4 * E - kf * C
    dDdt = k3 * C - km3 * D - kf * D
    dEdt = k4 * D - km4 * E - kf * E
    return dAdt, dBdt, dCdt, dDdt, dEdt


# With fixed k1, k2, k3, k4, km1, km2, km3, km4


sol = odeint(d4_rev_branched_dt, x0, t, args=(k1, km1, k2, km2, k3, km3, k4, km4, kf))

A_overlay = sol[:, 0] / sol[:, 0].mean()
B_overlay = sol[:, 1] / sol[:, 1].mean()
C_overlay = sol[:, 2] / sol[:, 2].mean()
D_overlay = sol[:, 3] / sol[:, 3].mean()
E_overlay = sol[:, 4] / sol[:, 4].mean()

fig, ax = plt.subplots(2, 1, figsize=(6, 2))
sns.lineplot(x=t, y=A_input, c="tab:blue", label="Input", linewidth=1.5, ax=ax[0])
sns.lineplot(x=t, y=A_overlay, c="tab:blue", label="A", linewidth=1.5, ax=ax[1])
sns.lineplot(x=t, y=B_overlay, c="tab:orange", label="B", linewidth=1.5, ax=ax[1])
sns.lineplot(x=t, y=C_overlay, c="tab:green", label="C", linewidth=1.5, ax=ax[1])
sns.lineplot(x=t, y=D_overlay, c="tab:red", label="D", linewidth=1.5, ax=ax[1])
sns.lineplot(x=t, y=E_overlay, c="tab:grey", label="E", linewidth=1.5, ax=ax[1])

# ax[0].set_ylim(0, 2)
ax[0].set_xlim(0, 300)
ax[1].set_ylim(0, 2)
ax[1].set_xlim(0, 300)

ax[0].set_ylabel("Input conc. (M)", fontsize=11)
ax[0].set_xlabel("Time (min)", fontsize=11)
ax[1].set_ylabel("Conc. / Avg.(Conc)", fontsize=11)
ax[1].set_xlabel("Time (min)", fontsize=11)

ax[0].legend(bbox_to_anchor=(1.02, 1.1), loc="upper left", fontsize=9)
ax[1].legend(bbox_to_anchor=(1.02, 1.1), loc="upper left", fontsize=9)

ax[0].set_facecolor("#F8F8FB")
ax[1].set_facecolor("#F8F8FB")
plt.subplots_adjust(
    left=None, bottom=None, right=None, top=1.2, wspace=None, hspace=0.55
)

# plt.savefig('Scheme_4_rev_reactions_branched.png', format='png', bbox_inches='tight', dpi=600)
```
